```python
import pandas as pd 
import numpy as np 
from sqlalchemy import create_engine
```

```python
#Load raw datasets 
df = pd.read_csv('Bank_Churn.csv')
print(f"Initial Shape: {df.shape}")
```

```python
# Data Cleaning & Validation
# Check for nulls and duplicate primary keys

print("Null Values:\n", df.isnull().sum())
print("Duplicates:", df.duplicated(subset=['CustomerId']).sum())
```

```python
# Strip any unexpected whitespace from text columns

df['Geography'] = df['Geography'].str.strip()
df['Gender'] = df['Gender'].str.strip()
df['Surname'] = df['Surname'].str.strip()
```

```python
# Round financial figures for data consistency

df['Balance'] = df['Balance'].round(2)
df['EstimatedSalary'] = df['EstimatedSalary'].round(2)
```
